Cell 1  → Import + Constants

```python
import random

# Winning combinations
WINNING_COMBINATIONS = [
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),
    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),
    (0, 4, 8),
    (2, 4, 6)
]

CORNERS = [0, 2, 6, 8]
SIDES = [1, 3, 5, 7]

AI = "O"
HUMAN = "X"
```

Cell 2  → Display Board

```python
def display_board(board):
    print()
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()
```

Cell 3  → Check Winne

```python
def check_winner(board, player):

    for a, b, c in WINNING_COMBINATIONS:

        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False
```

Cell 4  → Check Draw

```python
def is_draw(board):

    return all(cell != " " for cell in board)
```

Cell 5  → Find Winning Move

```python
def find_winning_move(board, player):

    for position in range(9):

        if board[position] == " ":

            # Temporarily place symbol
            board[position] = player

            # Check if this move wins
            if check_winner(board, player):

                # Undo move
                board[position] = " "

                return position

            # Undo move
            board[position] = " "

    return None
```

Cell 6  → Find Fork Move

```python
def find_fork_move(board, player):

    for position in range(9):

        if board[position] == " ":

            board[position] = player

            winning_moves = 0

            for next_position in range(9):

                if board[next_position] == " ":

                    board[next_position] = player

                    if check_winner(board, player):
                        winning_moves += 1

                    board[next_position] = " "

            board[position] = " "

            if winning_moves >= 2:
                return position

    return None
```

Cell 7  → Priority-Based AI

```python
def computer_move(board):

    move = find_winning_move(board, AI)

    if move is not None:
        print("Priority 1: Winning Move")
        return move


    move = find_winning_move(board, HUMAN)

    if move is not None:
        print("Priority 2: Blocking Opponent")
        return move

    move = find_fork_move(board, AI)

    if move is not None:
        print("Priority 3: Creating Fork")
        return move


    move = find_fork_move(board, HUMAN)

    if move is not None:
        print("Priority 4: Blocking Opponent's Fork")
        return move


    if board[4] == " ":
        print("Priority 5: Taking Center")
        return 4


    opposite_corners = {
        0: 8,
        8: 0,
        2: 6,
        6: 2
    }

    for corner in CORNERS:

        if board[corner] == HUMAN:

            opposite = opposite_corners[corner]

            if board[opposite] == " ":
                print("Priority 6: Taking Opposite Corner")
                return opposite


    available_corners = []

    for position in CORNERS:

        if board[position] == " ":
            available_corners.append(position)

    if available_corners:

        print("Priority 7: Taking Empty Corner")

        return random.choice(available_corners)

    available_sides = []

    for position in SIDES:

        if board[position] == " ":
            available_sides.append(position)

    if available_sides:

        print("Priority 8: Taking Empty Side")

        return random.choice(available_sides)

    return None
```

Cell 8  → Human Move

```python
def human_move(board):

    while True:

        try:

            position = int(input("Enter position (1-9): ")) - 1

            if position < 0 or position > 8:

                print("Enter a number between 1 and 9.")

            elif board[position] != " ":

                print("Position already occupied.")

            else:

                return position

        except ValueError:

            print("Please enter a valid number.")
```

Cell 9  → Main Game

```python
def play_game():

    board = [" "] * 9

    print("\n===== TIC-TAC-TOE =====")
    print("You = X")
    print("Computer = O")

    print("\nBoard Positions:")
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    current_player = HUMAN

    while True:

        display_board(board)

        # Human Turn
        if current_player == HUMAN:

            position = human_move(board)

            board[position] = HUMAN

            if check_winner(board, HUMAN):

                display_board(board)

                print("You Win!")

                break

            if is_draw(board):

                display_board(board)

                print("Game Draw!")

                break

            current_player = AI

        # Computer Turn
        else:

            print("\nComputer is thinking...")

            position = computer_move(board)

            if position is not None:

                board[position] = AI

                print("Computer selected position:", position + 1)

            if check_winner(board, AI):

                display_board(board)

                print("Computer Wins!")

                break

            if is_draw(board):

                display_board(board)

                print("Game Draw!")

                break

            current_player = HUMAN


play_game()
```
