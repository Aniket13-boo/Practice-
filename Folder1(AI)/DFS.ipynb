```python
from collections import deque
import matplotlib.pyplot as plt

V = int(input("Enter num of vertices: "))
E = int(input("Enter num of edges: "))

# Create adjacency list
adj = [[] for _ in range(V)]

print("Enter the edges (u v):")

edges = []

for i in range(E):
    u, v = map(int, input().split())

    adj[u].append(v)
    adj[v].append(u)

    edges.append((u, v))


plt.figure(figsize=(7, 6))

pos = {}

for i in range(V):
    angle = 2 * 3.14159 * i / V
    x = 5 * __import__('math').cos(angle)
    y = 5 * __import__('math').sin(angle)
    pos[i] = (x, y)

for u, v in edges:
    x1, y1 = pos[u]
    x2, y2 = pos[v]

    plt.plot([x1, x2], [y1, y2], 'k-')

for node in range(V):
    x, y = pos[node]

    plt.scatter(x, y, s=800)

    plt.text(x, y,
        str(node),
        ha='center',
        va='center',
        fontsize=12)

plt.title("Original Graph")
plt.axis("off")
plt.show()


start = int(input("Enter starting node: "))

visited = [False] * V
q = deque()

visited[start] = True
q.append(start)

print("BFS Traversal:", end=" ")

while q:

    node = q.popleft()

    print(node, end=" ")

    for neighbour in adj[node]:

        if not visited[neighbour]:

            visited[neighbour] = True
            q.append(neighbour)
```
