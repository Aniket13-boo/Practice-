```python
import matplotlib.pyplot as plt
import math


x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

dx = x2 - x1
dy = y2 - y1

length = max(abs(dx), abs(dy))
n = math.ceil(math.log2(length))

steps = 2 ** n

x_inc = dx / steps
y_inc = dy / steps

x = x1
y = y1

x_points = []
y_points = []

for i in range(steps + 1):
    x_points.append(int(x + 0.5))
    y_points.append(int(y + 0.5))

    x += x_inc
    y += y_inc
    
plt.scatter(x_points, y_points, color="#4d25c4")
plt.plot(x_points, y_points, color="#424bf5")
plt.title("Symmetrical DDA ")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.gca().set_aspect('equal')

plt.xticks(range(min(x_points), max(x_points) + 1))
plt.yticks(range(min(y_points), max(y_points) + 1))
for x, y in zip(x_points, y_points):
    plt.text(x, y + 0.1, f"({x},{y})", fontsize=9, ha='center')

plt.show()
```

```python
import matplotlib.pyplot as plt

x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

dx = x2 - x1
dy = y2 - y1

p = 2 * dy - dx

x = x1
y = y1

x_points = [x]
y_points = [y]

while x < x2:
    x += 1

    if p < 0:
        p += 2 * dy
    else:
        y += 1
        p += 2 * (dy - dx)

    x_points.append(x)
    y_points.append(y)

plt.scatter(x_points, y_points, color="#4d25c4")
plt.plot(x_points, y_points, color="#424bf5")
plt.title("Bresenham Line Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.xticks(range(min(x_points), max(x_points) + 1))
plt.yticks(range(min(y_points), max(y_points) + 1))

plt.grid(True)
plt.gca().set_aspect('equal')

for x, y in zip(x_points, y_points):
    plt.text(x, y + 0.1, f"({x},{y})", fontsize=9, ha='center')

plt.show()
```

```python
import matplotlib.pyplot as plt

xc = int(input("Enter center x: "))
yc = int(input("Enter center y: "))
r = int(input("Enter radius: "))

points = []


def plot(x, y):
    pts = [
        (xc + x, yc + y),
        (xc - x, yc + y),
        (xc + x, yc - y),
        (xc - x, yc - y),
        (xc + y, yc + x),
        (xc - y, yc + x),
        (xc + y, yc - x),
        (xc - y, yc - x)
    ]

    for p in pts:
        if p not in points:
            points.append(p)


x = 0
y = r
p = 1 - r

plot(x, y)

while x < y:
    x += 1

    if p < 0:
        p = p + 2 * x + 1
    else:
        y -= 1
        p = p + 2 * (x - y) + 1

    plot(x, y)


X = [i[0] for i in points]
Y = [i[1] for i in points]

plt.figure(figsize=(7,7))

plt.scatter(X, Y, s=50, color="black")

plt.grid(True)

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Midpoint Circle Drawing Algorithm")

plt.gca().set_aspect("equal")

plt.xticks(range(min(X)-2, max(X)+3))
plt.yticks(range(min(Y)-2, max(Y)+3))

plt.show()
```

```python
import matplotlib.pyplot as plt

# Input
xc = int(input("Enter x-coordinate of center: "))
yc = int(input("Enter y-coordinate of center: "))
rx = int(input("Enter x-radius (rx): "))
ry = int(input("Enter y-radius (ry): "))

x = 0
y = ry

x_points = []
y_points = []

algo_x = []
algo_y = []

def plot_points(xc, yc, x, y):
    x_points.extend([xc + x, xc - x, xc + x, xc - x])
    y_points.extend([yc + y, yc + y, yc - y, yc - y])

p1 = (ry * ry) - (rx * rx * ry) + (0.25 * rx * rx)
dx = 2 * ry * ry * x
dy = 2 * rx * rx * y

while dx < dy:

    algo_x.append(xc + x)
    algo_y.append(yc + y)

    plot_points(xc, yc, x, y)

    if p1 < 0:
        x += 1
        dx = 2 * ry * ry * x
        p1 = p1 + dx + ry * ry
    else:
        x += 1
        y -= 1
        dx = 2 * ry * ry * x
        dy = 2 * rx * rx * y
        p1 = p1 + dx - dy + ry * ry

p2 = (ry * ry * (x + 0.5) ** 2) + (rx * rx * (y - 1) ** 2) - (rx * rx * ry * ry)

while y >= 0:

    algo_x.append(xc + x)
    algo_y.append(yc + y)

    plot_points(xc, yc, x, y)

    if p2 > 0:
        y -= 1
        dy = 2 * rx * rx * y
        p2 = p2 - dy + rx * rx
    else:
        y -= 1
        x += 1
        dx = 2 * ry * ry * x
        dy = 2 * rx * rx * y
        p2 = p2 + dx - dy + rx * rx


plt.scatter(x_points, y_points, color="blue",marker='s',s=400, label="Symmetric Points")

plt.scatter(algo_x, algo_y, color="red",marker='s',s=400, label="Calculated Points")

for x, y in zip(algo_x, algo_y):
    plt.text(x + 0.2, y + 0.2, f"({x},{y})", fontsize=8)

plt.title("Midpoint Ellipse Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# plt.xticks(range(min(x_points), max(x_points) + 1))
# plt.yticks(range(min(y_points), max(y_points) + 1))
ax = plt.gca()

# Major ticks (labels only)
ax.set_xticks(range(min(x_points), max(x_points) + 1))
ax.set_yticks(range(min(y_points), max(y_points) + 1))

# Hide major tick marks if you don't want them
ax.tick_params(which='major', length=0)

# Minor ticks at .5 (for grid only)
ax.set_xticks([i + 0.5 for i in range(min(x_points)-1, max(x_points)+2)], minor=True)
ax.set_yticks([i + 0.5 for i in range(min(y_points)-1, max(y_points)+2)], minor=True)

# Grid only on minor ticks
ax.grid(False, which='major')
ax.grid(True, which='minor')

plt.xlim(min(x_points)-0.5, max(x_points)+0.5)
plt.ylim(min(y_points)-0.5, max(y_points)+0.5)

plt.gca().set_aspect('equal')
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

xc = int(input("Enter x-coordinate of center: "))
yc = int(input("Enter y-coordinate of center: "))
rx = int(input("Enter x-radius (rx): "))
ry = int(input("Enter y-radius (ry): "))

x = 0
y = ry

x_points = []
y_points = []

algo_x = []
algo_y = []

def plot_points(xc, yc, x, y):
    x_points.extend([xc + x, xc - x, xc + x, xc - x])
    y_points.extend([yc + y, yc + y, yc - y, yc - y])

d1 = (ry * ry) - (rx * rx * ry) + (rx * rx) / 4

dx = 2 * ry * ry * x
dy = 2 * rx * rx * y

while dx < dy:

    algo_x.append(xc + x)
    algo_y.append(yc + y)

    plot_points(xc, yc, x, y)

    if d1 < 0:
        x += 1
        dx = 2 * ry * ry * x
        d1 += dx + ry * ry
    else:
        x += 1
        y -= 1
        dx = 2 * ry * ry * x
        dy = 2 * rx * rx * y
        d1 += dx - dy + ry * ry

d2 = (ry * ry * (x + 0.5) * (x + 0.5)) + \
     (rx * rx * (y - 1) * (y - 1)) - \
     (rx * rx * ry * ry)

while y >= 0:

    algo_x.append(xc + x)
    algo_y.append(yc + y)

    plot_points(xc, yc, x, y)

    if d2 > 0:
        y -= 1
        dy = 2 * rx * rx * y
        d2 += rx * rx - dy
    else:
        x += 1
        y -= 1
        dx = 2 * ry * ry * x
        dy = 2 * rx * rx * y
        d2 += dx - dy + rx * rx

plt.scatter(x_points, y_points,
             marker='s',
            s=350,
            color="blue",
            label="Symmetric Points")

plt.scatter(algo_x, algo_y,
            color="red",
            marker='s',
            s=350,
            label="Calculated Points")

for x, y in zip(algo_x, algo_y):
    plt.text(x + 0.2, y + 0.2,
             f"({x},{y})",
             fontsize=8)

plt.title("Bresenham Ellipse Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")


ax = plt.gca()

ax.set_xticks(range(min(x_points), max(x_points) + 1))
ax.set_yticks(range(min(y_points), max(y_points) + 1))

# Hide major tick marks if you don't want them
ax.tick_params(which='major', length=0)

ax.set_xticks([i + 0.5 for i in range(min(x_points)-1, max(x_points)+2)], minor=True)
ax.set_yticks([i + 0.5 for i in range(min(y_points)-1, max(y_points)+2)], minor=True)

ax.grid(False, which='major')
ax.grid(True, which='minor')

plt.xlim(min(x_points)-0.5, max(x_points)+0.5)
plt.ylim(min(y_points)-0.5, max(y_points)+0.5)


plt.gca().set_aspect('equal')

plt.legend()

plt.show()
```
