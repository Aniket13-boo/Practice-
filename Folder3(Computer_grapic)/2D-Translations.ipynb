```python
import matplotlib.pyplot as plt

# Enter 3 points of triangle
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

x3 = float(input("Enter x3: "))
y3 = float(input("Enter y3: "))

points = [(x1, y1), (x2, y2), (x3, y3)]

print("\n1. Translation")
print("2. Scaling")
print("3. Composite (Translation + Scaling)")

choice = int(input("Enter your choice: "))

new_points = []

# Translation
if choice == 1:
    tx = float(input("Enter tx: "))
    ty = float(input("Enter ty: "))

    for x, y in points:
        new_points.append((x + tx, y + ty))

# Scaling
elif choice == 2:
    sx = float(input("Enter sx: "))
    sy = float(input("Enter sy: "))

    for x, y in points:
        new_points.append((x * sx, y * sy))

# Composite Transformation
elif choice == 3:
    tx = float(input("Enter tx: "))
    ty = float(input("Enter ty: "))

    sx = float(input("Enter sx: "))
    sy = float(input("Enter sy: "))

    for x, y in points:
        # Scaling first
        x_new = x * sx
        y_new = y * sy

        # Translation
        x_new = x_new + tx
        y_new = y_new + ty

        new_points.append((x_new, y_new))

else:
    print("Invalid choice")
    exit()

# Original triangle
old_x = [p[0] for p in points] + [points[0][0]]
old_y = [p[1] for p in points] + [points[0][1]]

# Transformed triangle
new_x = [p[0] for p in new_points] + [new_points[0][0]]
new_y = [p[1] for p in new_points] + [new_points[0][1]]

# Plot
plt.plot(old_x, old_y, marker='o', label="Original")
plt.plot(new_x, new_y, marker='s', label="Transformed")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Transformation")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import math

# Enter 3 points
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

x3 = float(input("Enter x3: "))
y3 = float(input("Enter y3: "))

points = [(x1, y1), (x2, y2), (x3, y3)]

print("\n1. Anticlockwise Rotation")
print("2. Clockwise Rotation")

choice = int(input("Enter your choice: "))
angle = float(input("Enter angle: "))

angle = math.radians(angle)

new_points = []

for x, y in points:

    if choice == 1:
        # Anticlockwise
        x_new = x * math.cos(angle) - y * math.sin(angle)
        y_new = x * math.sin(angle) + y * math.cos(angle)

    elif choice == 2:
        # Clockwise
        x_new = x * math.cos(angle) + y * math.sin(angle)
        y_new = -x * math.sin(angle) + y * math.cos(angle)

    else:
        print("Invalid choice")
        exit()

    new_points.append((x_new, y_new))

# Original triangle
old_x = [p[0] for p in points] + [points[0][0]]
old_y = [p[1] for p in points] + [points[0][1]]

# Rotated triangle
new_x = [p[0] for p in new_points] + [new_points[0][0]]
new_y = [p[1] for p in new_points] + [new_points[0][1]]

# Plot
plt.plot(old_x, old_y, marker='o', label="Original")
plt.plot(new_x, new_y, marker='s', label="Rotated")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Rotation of Triangle")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Number of vertices
n = int(input("Enter number of vertices: "))

# Input vertices
points = []

for i in range(n):
    x = float(input("Enter x" + str(i + 1) + ": "))
    y = float(input("Enter y" + str(i + 1) + ": "))
    points.append((x, y))

print("\n1. Reflection about X-axis")
print("2. Reflection about Y-axis")
print("3. Reflection about 45° line (y = x)")
print("4. Reflection about -45° line (y = -x)")

choice = int(input("Enter your choice: "))

new_points = []

# Reflection about X-axis
if choice == 1:
    for x, y in points:
        new_points.append((x, -y))

# Reflection about Y-axis
elif choice == 2:
    for x, y in points:
        new_points.append((-x, y))

# Reflection about y = x
elif choice == 3:
    for x, y in points:
        new_points.append((y, x))

# Reflection about y = -x
elif choice == 4:
    for x, y in points:
        new_points.append((-y, -x))

else:
    print("Invalid choice")
    exit()

# Original polygon
old_x = [p[0] for p in points] + [points[0][0]]
old_y = [p[1] for p in points] + [points[0][1]]

# Reflected polygon
new_x = [p[0] for p in new_points] + [new_points[0][0]]
new_y = [p[1] for p in new_points] + [new_points[0][1]]

# Plot
plt.plot(old_x, old_y, marker='o', label="Original")
plt.plot(new_x, new_y, marker='s', label="Reflected")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Reflection")
plt.grid(True)
plt.axis("equal")
# plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Number of vertices
n = int(input("Enter number of vertices: "))

# Input vertices
points = []

for i in range(n):
    x = float(input("Enter x" + str(i + 1) + ": "))
    y = float(input("Enter y" + str(i + 1) + ": "))
    points.append((x, y))

print("\n1. Shearing in X-direction")
print("2. Shearing in Y-direction")

choice = int(input("Enter your choice: "))
sh = float(input("Enter shear factor: "))

new_points = []

# Shearing
for x, y in points:

    if choice == 1:
        # X-direction
        x_new = x + sh * y
        y_new = y

    elif choice == 2:
        # Y-direction
        x_new = x
        y_new = y + sh * x

    else:
        print("Invalid choice")
        exit()

    new_points.append((x_new, y_new))

# Original polygon
old_x = [p[0] for p in points] + [points[0][0]]
old_y = [p[1] for p in points] + [points[0][1]]

# Sheared polygon
new_x = [p[0] for p in new_points] + [new_points[0][0]]
new_y = [p[1] for p in new_points] + [new_points[0][1]]

# Plot
plt.plot(old_x, old_y, marker='o', label="Original")
plt.plot(new_x, new_y, marker='s', label="Sheared")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Shearing")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```
